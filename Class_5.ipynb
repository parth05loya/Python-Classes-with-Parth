```python
#Collections
"""Hi, I am learing Python"""
from collections import Counter
sentence = "Hello world"
my_counter = Counter(sentence)
print(my_counter)
```

```python
"""a = {'value1': 1, 'value2': 2}"""
from collections import OrderedDict
ordered_dict = OrderedDict()
ordered_dict['alpha'] = 'bet'
ordered_dict['one'] = 'two'
ordered_dict['x'] = 'y'
print(ordered_dict)
```

```python
from itertools import product
prod = product([1,2], [5, 'a'])
print(list(prod))
```

```python
from itertools import permutations
perm = permutations([1,'a', 3, 'b'])
print(list(perm))
```

```python
from itertools import combinations
comb = combinations([1,2,'alpha', 4],2)
print(list(comb))
```
