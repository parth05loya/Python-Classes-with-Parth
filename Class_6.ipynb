```python
class Tutor:
  def __init__(self, name, timings):
    self.name = name
    self.timings = timings

python_tutor = Tutor("Parth", "6:00 AM") #Creating object of the class
music_tutor = Tutor("XYZ", "8:00 AM")
print(python_tutor.name, python_tutor.timings)
print(music_tutor.name, music_tutor.timings)
```

```python
#Attributes
class Tutor:
  student = "Rahul" #Class level attribute
  def __init__(self, name, timings, student): #name and timings are object attributes
    self.name = name
    self.timings = timings
    self.student = student

python_tutor = Tutor("Parth", "6:00 AM", "Lucky") #Creating object of the class
music_tutor = Tutor("XYZ", "8:00 AM", "Bhaskar")
print(python_tutor.name, python_tutor.timings, python_tutor.student)
print(music_tutor.name, music_tutor.timings, music_tutor.student)
print(Tutor.student) #Class attribute
```

```python
LEGB Rule: Local, Enclosed, Global, Builtin
```

```python
name = "madeline" ## Global
def enclosed_function():
  global name
  name = "Satish" ## Enclosing function
  def printer():
      name = "Aanya" ## Local
      print(name)
  print(name)
printer()
enclosed_function()
name
```

```python
# Sessions ,
# English, Public Speaking, Codings

# Assessments
# 96, 95, 100

class Sessions:
  #Constructor
  def __init__(self, session_name, assessment_score):
    self.name = session_name
    self.score = assessment_score

 #Method
  def get_avg(self):
    sum = 0
    for num in self.score:
      sum += num
    return sum/len(self.score)

session_first = Sessions("English", [90,95,98])
session_second = Sessions("Public Speaking", [100,85,98])
session_third = Sessions("Codings", [98,75,78])
print(session_first.name, session_first.score, session_first.get_avg())
print(session_second.name, session_second.score, session_second.get_avg())
print(session_third.name, session_third.score, session_third.get_avg())
```
