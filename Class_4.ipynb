This example takes a text string, loops through each character, and uses a dictionary to count how many vowels and consonants are present.

```python
string1 = "Hi , we are learning python"
dict1 = {"vowels": 0, "consonants": 0, "spaces": 0}
for char in string1:
  if char == 'a' or char == 'e' or char == 'i' or char == 'o' or char == 'u':
    dict1["vowels"]+=1
  elif char == ' ':
    dict1["spaces"]+= 1
  else:
    dict1["consonants"]+=1

print(dict1)
```

```python
opening_balance = 50
items_price = (12.5, 23.5, 10, 14) #Tuple
index = 0
while opening_balance > 0 and index < len(items_price):
  current_price = items_price[index]
  if opening_balance >= current_price:
    print("Purchased item no. {} for {}".format(index, current_price))
  else:
   print("Insufficient balance")
  opening_balance -= current_price
  index+=1

```
