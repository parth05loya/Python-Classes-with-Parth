```python
#zip combines two list into a tuple
list1 = ['a', 'b', 'c', 'd', 'e']
list2 = [1,2,3,4,5]
for (a,b) in zip(list1, list2):
  print((a,b))
```

```python
String = 'Print every word in this sentence that has even number of letters'
Example output: word <-- even
                this <-- even
```

```python
String = 'Print every word in this sentence that has even number of letters'
c = String.split()
for i in c:
  if len(i)%2 == 0:
    print(i + " <-- even")
```
